````python
# Initialize the message history lists before the loop
examiner_messages = []
student_messages = []

# Initial instruction to start the chain
question = "You are an Examiner. Start the exam by asking a grade-8 math question."

for i in range(1, 4):
    print(f"\n=================== ROUND {i} ===================")  #will run for 3 rounds starting from 1 and end at 3 
    
    # 1. EXAMINER TURN
    examiner_messages.append({"role": "user", "content": question})  #will append the examiner question to the content question
    response = client.chat.completions.create(
        model="gemini-3.5-flash",
        messages=examiner_messages)
    examiner_reply = response.choices[0].message.content
    examiner_messages.append({"role": "assistant", "content": examiner_reply})   # here assistant work as history for the answers given by the AI 
    print(f"\n[EXAMINER]:\n{examiner_reply}")

    # 2. STUDENT TURN
    question = f"You are a student . Answer this question step-by-step:\n\n{examiner_reply}"  #student will give answer to the examiner question step by step
    student_messages.append({"role": "user", "content": question})
    response = client.chat.completions.create(
        model="gemini-3.5-flash", 
        messages=student_messages)
    student_reply = response.choices[0].message.content
    student_messages.append({"role": "assistant", "content": student_reply})         
    print(f"\n[STUDENT]:\n{student_reply}")

    # 3. PREPARE INPUT FOR NEXT ROUND
    # Pass the student's reply back to the examiner to score and ask a harder question
    question = f"You are the Examiner. Score this student's answer out of 3 and ask a harder grade-8 math question:\n\n{student_reply}"
````

````python
# Create a Pydantic model for the Evaluation

from pydantic import BaseModel 

# Pydantic is used for data validation and settings management using Python type annotations.
# BaseModel is the core class used to define structured data schemas with automatic data type validation and parsing.
class Evaluation(BaseModel):

    # Field to hold a boolean flag indicating whether the output meets evaluation criteria (True/False)
    is_acceptable: bool

    # Field to provide textual feedback explaining why the output passed or failed verification
    feedback: str
````

````python
# Construct the evaluator system prompt by defining the identity and role guidelines for the evaluation model.

evaluator_system_prompt = f"You are an evaluator that decides whether a response to a question is acceptable. \
You are provided with a conversation between a User and an Agent. Your task is to decide whether the Agent's latest response is acceptable quality. \
The Agent is playing the role of {name} and is representing {name} on their website. \
The Agent has been instructed to be professional and engaging, as if talking to a potential client or future employer who came across the website. \
The Agent has been provided with context on {name} in the form of their LinkedIn details. Here's the information:"

# Append the extracted CV/LinkedIn details to the evaluator prompt as structural context.
evaluator_system_prompt += f"\n\n## LinkedIn Profile:\n{linkedin}\n\n"

# Add final instruction specifying the required output criteria (acceptability decision and feedback).
evaluator_system_prompt += f"With this context, please evaluate the latest response, replying with whether the response is acceptable and your feedback."
````

````python
#function with 3 parameters. All parameters used inside function

def evaluator_user_prompt(reply, message, history):

    # Format the complete conversation history context between the User and the Agent.
    user_prompt = f"Here's the conversation between the User and the Agent: \n\n{history}\n\n"
    user_prompt += f"Here's the latest message from the User: \n\n{message}\n\n"
    user_prompt += f"Here's the latest response from the Agent: \n\n{reply}\n\n"
    user_prompt += "Please evaluate the response, replying with whether it is acceptable and your feedback."
   
   # Return the assembled evaluation prompt string.
    return user_prompt
````

````python
def evaluate(reply, message, history) -> Evaluation:

    messages = [{"role": "system", "content": evaluator_system_prompt}] + [{"role": "user", "content": evaluator_user_prompt(reply, message, history)}]
    response = openai.beta.chat.completions.parse(model="gpt-4o-mini", messages=messages, response_format=Evaluation)
    return response.choices[0].message.parsed 

    # . beta is used to force the answer into a specific structure
    # .parsed used instead of just raw text in .content — already includes a .parsed attribute containing a real Python object instead of digging through a JSON string.
````

````python
def rerun(reply, message, history, feedback):

    # Construct an augmented system prompt that includes critique from the failed evaluation step.
    updated_system_prompt = system_prompt + "\n\n## Previous answer rejected\nYou just tried to reply, but the quality control rejected your reply\n"
    updated_system_prompt += f"## Your attempted answer:\n{reply}\n\n"
    updated_system_prompt += f"## Reason for rejection:\n{feedback}\n\n"
   
   # Reassemble the messages payload with the updated system prompt, full history, and current user query.
    messages = [{"role": "system", "content": updated_system_prompt}] + history + [{"role": "user", "content": message}]
   
   # Send the request back to the OpenAI API so the agent can self-correct based on the evaluation feedback.
    response = openai.chat.completions.create(model="gpt-4o-mini", messages=messages)
   
   # Return the newly generated self-corrected response.
    return response.choices[0].message.content
````

````python
def chat(message, history):

    # Conditionally modify the system prompt to introduce a deliberate policy test (Pig Latin) if the query contains "patent"
    
    if "patent" in message:
        system = system_prompt + "\n\nEverything in your reply needs to be in pig latin - \
              it is mandatory that you respond only and entirely in pig latin"
    else:
        system = system_prompt

    # Assemble the API payload using the updated system prompt, conversation history, and user input    
    messages = [{"role": "system", "content": system}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model="gpt-4o-mini", messages=messages)
    reply =response.choices[0].message.content

# Validate the reply using our structured Pydantic evaluation pipeline
    evaluation = evaluate(reply, message, history)
    
# Process quality gate logic: return original if approved, or invoke self-correction via rerun() if rejected
      
    if evaluation.is_acceptable:
        print("Passed evaluation - returning reply")
    else:
        print("Failed evaluation - retrying")
        print(evaluation.feedback)
        reply = rerun(reply, message, history, evaluation.feedback)       
    return reply
````

````python
# Launch the updated Gradio chat interface with the built-in evaluation and self-correction pipeline.
gr.ChatInterface(chat).launch()
````

````python
# For TELEGRAM

# Retrieve Telegram credentials from environment variables loaded via dotenv.
TELEGRAM_BOT_TOKEN = os.getenv("TELEGRAM_BOT_TOKEN")
TELEGRAM_CHAT_ID = os.getenv("TELEGRAM_CHAT_ID")

# Verify the Telegram Bot token exists and print a safe confirmation snippet.
if TELEGRAM_BOT_TOKEN:
    print(f"Telegram Bot token found and starts with {TELEGRAM_BOT_TOKEN[0]}")
else:
    print("Telegram Bot token not found")

# Verify the Telegram Chat ID exists and print a safe confirmation snippet.
if TELEGRAM_CHAT_ID :
    print(f"Telegram chat id found and starts with {TELEGRAM_CHAT_ID [0]}")
else:
    print("Telegram chat id not found")
````

````python
# def push(message):
#     print(f"Push: {message}")
#     payload = {"user":Telegram_bot , "token":TELEGRAM_BOT_TOKEN , "message": message}
#     requests.post(URL, data=payload)

# Function named `push` that sends messages via Telegram Bot
def push(message):
    # Prints the message locally to the console/logs
    print(f"Push: {message}", flush=True)

    # Check if credentials are loaded
    if not TELEGRAM_BOT_TOKEN or not TELEGRAM_CHAT_ID:
        print("Telegram credentials missing. Cannot send message.")
        return

    # Telegram API URL for sending messages
    url = f"https://api.telegram.org/bot{TELEGRAM_BOT_TOKEN}/sendMessage"

    # Data payload required by the Telegram Bot API
    payload = {
        "chat_id": TELEGRAM_CHAT_ID,
        "text": message
    }

    # Send the HTTP POST request to Telegram API
    try:
        response = requests.post(url, json=payload)
        response.raise_for_status()
    except Exception as e:
        print(f"Failed to send Telegram notification: {e}")
````

````python
# This is a more elegant way that avoids the IF statement.

# Function that processes all tool calls returned by the LLM
def handle_tool_calls(tool_calls):

    # Create an empty list to store results from each tool execution
    results = []

    # Loop through each tool call suggested by the LLM
    for tool_call in tool_calls:

        # Extract the tool/function name
        # Example: "record_user_details"
        tool_name = tool_call.function.name

        # Convert the JSON string of arguments into a Python dictionary
        # Example:
        # '{"email":"john@example.com","name":"John"}'
        # becomes:
        # {"email":"john@example.com","name":"John"}
        arguments = json.loads(tool_call.function.arguments)

        # Print the name of the tool being executed
        # flush=True ensures it appears immediately in the console
        print(f"Tool called: {tool_name}", flush=True)

        # Look for a global function with the same name as tool_name
        # Example:
        # globals().get("record_user_details")
        # returns the actual function object:
        # <function record_user_details>
        tool = globals().get(tool_name)

        # If the function exists:
        #   tool(**arguments)
        #   calls the function with unpacked arguments
        #
        # Example:
        # arguments = {"email":"john@example.com","name":"John"}
        #
        # tool(**arguments)
        #
        # becomes:
        # record_user_details(
        #     email="john@example.com",
        #     name="John"
        # )
        #
        # If the function doesn't exist, return an empty dictionary
        result = tool(**arguments) if tool else {}

        # Add the tool result in the format expected by the OpenAI API
        results.append({

            # Indicates this message came from a tool
            "role": "tool",

            # Convert Python dictionary result into JSON string
            # Example:
            # {"recorded":"ok"}
            # becomes:
            # '{"recorded":"ok"}'
            "content": json.dumps(result),

            # Links this result back to the original tool call
            "tool_call_id": tool_call.id
        })

    # Return all tool results to be sent back to the LLM
    return results
````

````python
# Create a reader object for the target PDF file (a CV document).
reader = PdfReader("me/Bhavisha_Tak_CV.pdf")

# Initialize an empty string variable to store all collected text.
linkedin = ""

# Loop through every page in the loaded PDF document.
for page in reader.pages:

    # Extract plain text content from the current page.
    text = page.extract_text()

    # Check if text was successfully extracted from the page.
    if text:
        # Append the extracted page text to our combined string variable.
        linkedin += text

name = "Bhavisha Tak"
````

````python
# System prompt is different from user prompt as it has overall instructions that sets context for task at hand and 
# user prompt is actual question

system_prompt = f"""You are acting as {name}. You are answering questions on {name}'s personal website, \
particularly questions related to {name}'s career, background, skills, and experience. \
Your responsibility is to represent {name} faithfully and professionally to visitors — potential clients, \
recruiters, or future employers — who come across the site.

You will be given {name}'s CV/profile below. This is your ONLY source of truth about {name}.

Rules you must follow:
1. Answer questions using ONLY the information present in the profile provided below.
2. Never guess or invent details. If a question asks for something not covered in the profile \
(e.g. personal opinions, unlisted skills, salary expectations, availability, or anything not \
explicitly stated), you must NOT say "I don't have that information" or answer it directly — \
instead follow Rule 7 below (ask for their email first).
3. Do not fabricate job titles, dates, companies, skills, or achievements that are not in the profile.
4. Stay in character as {name} at all times — respond in first person ("I worked at...", "My experience includes...").
5. Keep responses professional, warm, and engaging, as if speaking directly with someone interested in \
{name}'s work.
6. If asked something completely unrelated to {name}'s career/background (e.g. general trivia, coding help \
for the visitor, etc.), politely redirect the conversation back to {name}'s professional background.
7. CRITICAL: If you cannot answer a question because the information is not in the profile, do NOT \
answer it, do NOT say "I don't have that information", and do NOT make excuses. Your VERY FIRST step \
MUST be to politely ask the visitor for their email address so {name} can follow up personally — do not \
call any tool on this turn. Once the visitor responds on their next message (with an email, or by \
declining), call record_unknown_question with the original question and their email (use 'not provided' \
if they decline or don't give one).
8. If the visitor shares their email address or asks to be contacted, or expresses strong interest in \
getting in touch, you MUST call the record_user_details tool with their email (and name/notes if given).

## Profile:
{linkedin}

With this context, please chat with the user, always staying in character as {name}.
"""
````

````python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    done = False
    while not done: # keep calling openai until done is true that is finish reason is not a tool call

        # This is the call to the LLM - see that we pass in the tools json
        # Here tools is structured json containing two functions and tell LLM these things you can do
        # response could be for user or may be tool LLM needs to use
        response = openai.chat.completions.create(model="gpt-4o-mini", messages=messages, tools=tools)

        finish_reason = response.choices[0].finish_reason
        
        # If the LLM wants to call a tool, we do that!
        if finish_reason=="tool_calls":
            message = response.choices[0].message
            tool_calls = message.tool_calls
            results = handle_tool_calls(tool_calls) # call handle_tool_calls function to give LLM access of tool it asked for
            messages.append(message)
            messages.extend(results)
        else:
            done = True
    return response.choices[0].message.content

    # keep running a loop until finish reason is not a tool call
````
